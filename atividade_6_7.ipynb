#2- Contador de número

```python
# Solicita o número ao usuário
numero = int(input("Digite um número inteiro positivo: "))

# Converte para string e calcula o tamanho
quantidade_digitos = len(str(numero))

print(f"O número {numero} possui {quantidade_digitos} dígitos.")
```

#3 - Cofrinho

```python
# Configuração da meta
meta = 100.00
total_acumulado = 0.0
meses = 0

# Entrada de dados do usuário
valor_deposito = float(input("Informe o valor do depósito mensal (R$): "))

# Estrutura de repetição para calcular os meses
while total_acumulado < meta:
    total_acumulado += valor_deposito
    meses += 1

# Apresentação dos resultados
print("\n--- Resultado ---")
print(f"Total acumulado: R$ {total_acumulado:.2f}")
print(f"Meses necessários: {meses}")
print(f"A meta foi atingida no mês: {meses}")
```

#4- Caixa Registradora

```python
subtotal_geral = 0

while True:
    preco = float(input("Digite o preço do produto (ou 0 para sair): "))

    if preco == 0:
        break

    quantidade = int(input("Digite a quantidade: "))

    # Cálculo do subtotal do item atual
    subtotal_item = preco * quantidade
    subtotal_geral += subtotal_item
    print(f"Subtotal deste item: R$ {subtotal_item:.2f}\n")

# Lógica de definição de desconto com base no subtotal acumulado
if subtotal_geral >= 500:
    taxa_desconto = 0.15 # 15%
elif subtotal_geral >= 300:
    taxa_desconto = 0.10 # 10%
elif subtotal_geral >= 100:
    taxa_desconto = 0.05 # 5%
else:
    taxa_desconto = 0.00 # 0%

# Cálculos finais
valor_desconto = subtotal_geral * taxa_desconto
valor_final = subtotal_geral - valor_desconto

# Exibição dos resultados finais
print("="*30)
print("       RESUMO DA COMPRA       ")
print("="*30)
print(f"Subtotal: R$ {subtotal_geral:.2f}")
print(f"Desconto aplicado: {taxa_desconto * 100:.0f}% (R$ {valor_desconto:.2f})")
print(f"Valor Final: R$ {valor_final:.2f}")
print("="*30)
```
