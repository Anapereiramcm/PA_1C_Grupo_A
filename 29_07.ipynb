```python
#Escrever um programa que identifica se o número é positivo ou negativo.

cu = int(input("Digite um número inteiro:  "))

if cu > 0:
  print("")
  print(f"O número {cu} é positivo.")

elif cu < 0:
  print("")
  print(f"O número {cu} é negativo.")

elif cu == 0:
  print("")
  print(f"O número {cu} é um número neutro. Por isso, ele não")
  print("se encaixa na classificação positiva nem negativa.")
```

```python
#Escrever um programa que recebe um número menor que 20 e imprime os números pares

a = int(input("Digite um número menor que 20 e maior que 0:  "))

if a < 20 and a > 0:
  print("")
  print(f"Os números pares de 0 a {a} são:")
  for i in range (a + 1):
    if i % 2 == 0:
      print(i)


if a < 0:
  print("")
  print(f"O número {a} é negativo. Programa encerrado.")

elif a > 20:
  print("")
  print(f"O número {a} é maior que 20. Programa encerrado.")

```
