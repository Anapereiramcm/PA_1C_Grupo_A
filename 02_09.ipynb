```python
#4- Crie uma lista contendo 5 notas de um aluno. Ex.: notas= [7.5, 8.0, 6.5, 9.0, 8.5]

#Calcule: a soma das notas, a média, a maior nota e a menor nota.
#Desafio; determine se o aluno foi aprovado considerando média maior ou igual a 7

notas= [7.5, 8.0, 6.5, 9.0, 8.5]
aprovados = []
reprovados=[]
print("Soma das notas: ")

print(sum(notas))
print("")
print("Média das notas:")
print(sum(notas)/len(notas))
print("")
print("Maior nota:")
print(max(notas))
print("")
print("Menor nota:")
print(min(notas))
print("")
print("Aprovados:")
for notas in notas:
  if notas >= 7:
    aprovados.append(notas)
  elif notas < 7:
    reprovados.append(notas)


print(aprovados)
print("")
print("Reprovados:")
print(reprovados)
```

```python
#9- Crie uma lista contendo 5 produtos. Para cada produto, armazene:
#Nome, preço.
#Uma possinilidade é: [
    #["Arroz", 25.90],
    #["Feijão", 8.50]
    #["Macarrão", 6.90]
    #["Leite, 5.50"]
    #["Café", 15.00]
#Faça um programa que mostre todos os produtos e seus preços
#Que calcule o valor total da comprasa

#Encontre o produto mais caro
#Encontre o produto mais barato.
#Desafio: Permita que o usuário informe um produto e verifique se ele está cadastrado.

# Lista com 5 produtos e seus preços
produtos = [
    ["Arroz", 25.90],
    ["Feijão", 8.50],
    ["Macarrão", 6.90],
    ["Leite", 5.50],
    ["Café", 15.00]
]

print("Produtos")
total = 0
for item in produtos:
    nome = item[0]
    preco = item[1]
    print(f"- {nome}: R$ {preco:.2f}")
    total += preco

print(f"\nValor total da compra: R$ {total:.2f}")

caro = max(produtos, key=lambda x: x[1])
barato = min(produtos, key=lambda x: x[1])

print(f"\nProduto mais caro: {caro[0]} (R$ {caro[1]:.2f})")
print(f"Produto mais barato: {barato[0]} (R$ {barato[1]:.2f})")
obunga= input("\nDigite o nome de um produto para verificar o cadastro: ").strip()
encontrado = False

for item in produtos:
    if item[0].lower() == obunga.lower():
        print(f"O produto '{item[0]}' está cadastrado e custa R$ {item[1]:.2f}.")
        encontrado = True
        break

if not encontrado:
    print(f"O produto '{obunga}' não está cadastrado.")


```

```python
#10- Crie um programa que permita cadastrar as notas de vários alunos em uma lista.
#O programa deve: Perguntar quantos alunos serão cadastrados
#Para cada aluno, pedir: nome, nota
#Armazenar as informações em uma lista
#Ao final, mostrar: Todos os alunos e suas notas
#A média da turma, o aluno com maior nota
#O aluno com menor nota, quais alunos foram aprovados (nota >= 7)
#quais alunos foram reprovados (nota<7)
#Desafio extra: permita que o usuário pesquise um aluno pelo nome e veja sua nota.
alunos = []
qtd_alunos = int(input("Quantos alunos serão cadastrados? "))
for i in range(qtd_alunos):
    print(f"\n--- Cadastro do {i+1}º Aluno ---")
    nome = input("Nome: ").strip()
    nota = float(input("Nota: "))

    alunos.append({"nome": nome, "nota": nota})

if qtd_alunos > 0:
    soma_notas = 0
    maior_aluno = alunos[0]
    menor_aluno = alunos[0]
    aprovados = []
    reprovados = []

    print("\n================ STATS DA TURMA ================")
    print("Alunos cadastrados e suas notas:")

    for aluno in alunos:
        print(f"- {aluno['nome']}: {aluno['nota']}")

        soma_notas += aluno['nota']

        if aluno['nota'] > maior_aluno['nota']:
            maior_aluno = aluno

        if aluno['nota'] < menor_aluno['nota']:
            menor_aluno = aluno

        if aluno['nota'] >= 7.0:
            aprovados.append(aluno['nome'])
        else:
            reprovados.append(aluno['nome'])

    media = soma_notas / qtd_alunos

    print(f"\nMédia da turma: {media:.2f}")
    print(f"Maior nota: {maior_aluno['nome']} ({maior_aluno['nota']})")
    print(f"Menor nota: {menor_aluno['nome']} ({menor_aluno['nota']})")
    print(f"Alunos aprovados: {', '.join(aprovados) if aprovados else 'Nenhum'}")
    print(f"Alunos reprovados: {', '.join(reprovados) if reprovados else 'Nenhum'}")
    print("\n================ SISTEMA DE PESQUISA ================")
    while True:
        busca = input("\nDigite o nome do aluno para pesquisar (ou 'sair' para fechar): ").strip()

        if busca.lower() == 'sair':
            print("Programa encerrado.")
            break

        encontrado = False
        for aluno in alunos:
            if aluno['nome'].lower() == busca.lower():
                print(f"Aluno encontrado! A nota de {aluno['nome']} é {aluno['nota']}.")
                encontrado = True
                break

        if not encontrado:
            print("Aluno não encontrado no sistema.")
else:
    print("Nenhum aluno cadastrado.")

```
